# MNIST Dataset Exploration

This notebook presents an exploratory analysis of the **MNIST dataset of handwritten digits** using TensorFlow and Keras. This forms **Step 1** of the Handwritten Digit Recognition assignment project.

## 1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

print(f"TensorFlow Version: {tf.__version__}")
```

## 2. Load Dataset

We load the standard MNIST dataset directly using Keras built-in utilities (`tf.keras.datasets.mnist.load_data()`).

```python
# Load train and test sets from TensorFlow / Keras
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("MNIST dataset successfully loaded.")
```

## 3. Dataset Dimensions

```python
print(f"Training image shape : {x_train.shape}")
print(f"Training label shape : {y_train.shape}")
print(f"Testing image shape  : {x_test.shape}")
print(f"Testing label shape  : {y_test.shape}")
```

## 4. Visualize Sample Images

We visualize several sample images from the training dataset along with their ground-truth labels.

```python
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(f"Label: {y_train[i]}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

## 5. Pixel Value Inspection

Before preprocessing, we check the minimum and maximum raw pixel intensity values.

```python
print(f"Minimum pixel value (raw): {x_train.min()}")
print(f"Maximum pixel value (raw): {x_train.max()}")
print(f"Data type of raw images: {x_train.dtype}")
```

## 6. Data Normalization

### What is Normalization?
**Normalization** is the data preprocessing technique of rescaling numerical values into a standard range, typically between `0.0` and `1.0`.

### Why do we normalize image pixel values?
1. **Numerical Stability:** In raw MNIST images, pixel intensities range from `0` (black background) to `255` (white foreground). Large input values can lead to large gradient updates during optimization, causing instability or slow convergence.
2. **Faster Gradient Descent:** Scaling inputs to $[0, 1]$ ensures smooth and faster gradient descent convergence during neural network backpropagation.
3. **Uniform Scale:** Prevents any specific high-value pixels from artificially dominating model weight updates.

```python
# Rescale pixel values from [0, 255] to [0.0, 1.0]
x_train_norm = x_train.astype('float32') / 255.0
x_test_norm = x_test.astype('float32') / 255.0

print(f"Minimum pixel value (normalized): {x_train_norm.min():.2f}")
print(f"Maximum pixel value (normalized): {x_train_norm.max():.2f}")
print(f"Data type after normalization: {x_train_norm.dtype}")
```

## 7. Label Verification

We inspect the target labels to confirm that all 10 digit classes (0–9) are represented.

```python
unique_labels, counts = np.unique(y_train, return_counts=True)
print("Unique target labels found:", unique_labels)
print("\nClass distribution in training set:")
for label, count in zip(unique_labels, counts):
    print(f"  Digit {label}: {count} samples")
```

## Observations

### Key Insights from MNIST Exploration:
1. **Dataset Size & Split:** The MNIST dataset contains **60,000 training samples** and **10,000 testing samples**.
2. **Image Dimensions:** Each image is a single-channel grayscale image of resolution **$28 \times 28$ pixels**.
3. **Pixel Representation:** Raw images contain integer pixel intensities ranging from **0** to **255** (`uint8`).
4. **Data Normalization:** Dividing by `255.0` rescales pixel values to `float32` in range **$[0.0, 1.0]$**, optimizing the dataset for neural network training.
5. **Target Class Balance:** The labels contain digits **0 through 9**, with a balanced class distribution across training samples (roughly 5,400 to 6,700 images per digit category).
